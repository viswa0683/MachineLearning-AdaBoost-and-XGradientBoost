```python
# Pandas Library is imported into the file Multiple Regression Exercise 1
import pandas as pd
```

```python
# Variable l_dataset
# assign the read_csv function from pandas into the variable l_dataset
l_dataset = pd.read_csv("50_Startups.csv")
```

```python
l_dataset
```

```python
# to convert text (categorical data) into numbers (0 and 1).
# converts categories into separate columns.
l_dataset2 =pd.get_dummies(l_dataset)
```

```python
l_dataset2
```

```python
l_dataset3 = pd.get_dummies(l_dataset, dtype=int, drop_first=True)
```

```python
l_dataset3
```

```python
# it print  all the column heading in array list
l_dataset3.columns
```

```python
# Now we split the input and output columns to the variable
independent=l_dataset3[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
```

```python
independent
```

```python
dependent = l_dataset3[["Profit"]]
```

```python
dependent
```

```python
# Now we going to split , Training Set and Test Set
# Import the sklearn library, module of model_selection and train_test_split function into the Model creation
from sklearn.model_selection import train_test_split
```

```python
train_test_split (independent, dependent, test_size=0.30,random_state=0)
```

```python
X_train, X_test, y_train, y_test=train_test_split (independent, dependent, test_size=0.30,random_state=0)
```

```python
X_train
```

```python
y_train
```

```python
# Adaboost Regressor Model creation
# Model creation
# loss{‘squared_error’, ‘absolute_error’, ‘huber’, ‘quantile’}, default=’squared_error’
from sklearn.ensemble import GradientBoostingRegressor
l_reg = GradientBoostingRegressor(loss='squared_error',random_state=None)

# Train the model 
l_reg.fit (X_train, y_train)
```

```python
# Test the model	
y_pred = l_reg.predict (X_test)
```

```python
y_pred
```

```python
l_reg.score(X_train, y_train)
```

```python
from sklearn.metrics import r2_score
l_score=r2_score (y_test, y_pred)
l_score
```

```python
import pickle
filename="XgradientBoostingRegressor_file.sav"
pickle.dump(l_reg,open (filename,'wb'))
```
